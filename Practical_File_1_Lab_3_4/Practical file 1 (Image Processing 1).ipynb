## Program 3 – Colour Image to Grayscale Conversion

**Objective:** To convert a colour image into grayscale using OpenCV and save the converted image.

```python
import cv2
import matplotlib.pyplot as plt

# Load the colour image
img = cv2.imread("scene.jpg")

if img is None:
    print("Error: Image not found!")
else:
    # Convert colour image to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Save the grayscale image
    cv2.imwrite("output_gray.jpg", gray)

    print("Image converted to grayscale successfully!")
    print("Grayscale image saved as output_gray.jpg")

    # Display original and grayscale images
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Colour Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.show()
```

## Program 5 – Image Resizing

**Objective:** To resize an image to 50% of its original dimensions using OpenCV.

```python
import cv2
import matplotlib.pyplot as plt

# Read the original image
img = cv2.imread("scene.jpg")

if img is None:
    print("Error: Image not found!")
else:
    # Get original dimensions
    height, width = img.shape[:2]

    # Resize image to 50%
    new_width = int(width * 0.5)
    new_height = int(height * 0.5)

    resized_img = cv2.resize(img, (new_width, new_height))

    # Convert images from BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    resized_rgb = cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB)

    # Display dimensions
    print("Original Image Size:", width, "x", height)
    print("Resized Image Size:", new_width, "x", new_height)

    # Display both images
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(resized_rgb)
    plt.title("Resized Image (50%)")
    plt.axis("off")

    plt.show()
```

```python
cv2.imwrite("resized_image.jpg", resized_img)
print("Resized image saved successfully!")
```

## Program 6 – Image Cropping

**Objective:** To crop the central 200 × 200 pixel area from an image using OpenCV.

```python
import cv2
import matplotlib.pyplot as plt

# Read the original image
img = cv2.imread("scene.jpg")

if img is None:
    print("Error: Image not found!")
else:
    height, width = img.shape[:2]

    # Check image dimensions
    if height < 200 or width < 200:
        print("Image is too small for 200 x 200 cropping.")
    else:
        # Calculate the centre coordinates
        center_x = width // 2
        center_y = height // 2

        # Calculate cropping boundaries
        x1 = center_x - 100
        x2 = center_x + 100
        y1 = center_y - 100
        y2 = center_y + 100

        # Crop the central region
        cropped_img = img[y1:y2, x1:x2]

        # Display original and cropped images
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        cropped_rgb = cv2.cvtColor(cropped_img, cv2.COLOR_BGR2RGB)

        print("Original Image Dimensions:", img.shape)
        print("Cropped Image Dimensions:", cropped_img.shape)

        plt.figure(figsize=(12, 5))

        plt.subplot(1, 2, 1)
        plt.imshow(img_rgb)
        plt.title("Original Image")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(cropped_rgb)
        plt.title("Central 200 x 200 Cropped Image")
        plt.axis("off")

        plt.show()

        # Save cropped image
        cv2.imwrite("cropped_image.jpg", cropped_img)
        print("Cropped image saved successfully!")
```

## Program 7 – Image Rotation

**Objective:** To rotate an image by 90 degrees clockwise using OpenCV.

```python
import cv2
import matplotlib.pyplot as plt

# Read the original image
img = cv2.imread("scene.jpg")

if img is None:
    print("Error: Image not found!")
else:
    # Rotate image 90 degrees clockwise
    rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # Convert images from BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    rotated_rgb = cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB)

    # Display dimensions
    print("Original Image Dimensions:", img.shape)
    print("Rotated Image Dimensions:", rotated_img.shape)

    # Display both images
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(rotated_rgb)
    plt.title("Rotated Image - 90 Degrees Clockwise")
    plt.axis("off")

    plt.show()

    # Save rotated image
    cv2.imwrite("rotated_image.jpg", rotated_img)
    print("Rotated image saved successfully!")
```

## Program 8 – Horizontal Image Flipping

**Objective:** To flip an image horizontally using OpenCV.

```python
import cv2
import matplotlib.pyplot as plt

# Read the original image
img = cv2.imread("car.jpg")

if img is None:
    print("Error: Image not found!")
else:
    # Flip image horizontally
    flipped_img = cv2.flip(img, 1)

    # Convert images from BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    flipped_rgb = cv2.cvtColor(flipped_img, cv2.COLOR_BGR2RGB)

    # Display both images
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Car Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(flipped_rgb)
    plt.title("Horizontally Flipped Image")
    plt.axis("off")

    plt.show()

    # Save flipped image
    cv2.imwrite("flipped_car.jpg", flipped_img)

    print("Image flipped horizontally successfully!")
    print("Flipped image saved as flipped_car.jpg")
```

## Program 9 – RGB to HSV Colour Space Conversion

**Objective:** To convert a colour image into HSV and display the Hue, Saturation and Value channels separately.

```python
import cv2
import matplotlib.pyplot as plt

# Read the original image
img = cv2.imread("scene.jpg")

if img is None:
    print("Error: Image not found!")
else:
    # Convert BGR image to HSV
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    # Split HSV into three channels
    h, s, v = cv2.split(hsv)

    # Display the three channels
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(h, cmap="gray")
    plt.title("Hue Channel")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(s, cmap="gray")
    plt.title("Saturation Channel")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(v, cmap="gray")
    plt.title("Value Channel")
    plt.axis("off")

    plt.show()

    print("HSV conversion completed successfully!")

    cv2.imwrite("hue.jpg", h)
cv2.imwrite("saturation.jpg", s)
cv2.imwrite("value.jpg", v)

print("All three HSV channels saved successfully!")
```

## Program 10 – Mean Squared Error (MSE)

```python
import cv2
import matplotlib.pyplot as plt
from skimage.metrics import mean_squared_error

# Read both images in grayscale
img1 = cv2.imread("img1.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("img2.jpg", cv2.IMREAD_GRAYSCALE)

if img1 is None or img2 is None:
    print("Error: Please check whether both images are uploaded correctly.")

else:
    # Match image dimensions if they are different
    if img1.shape != img2.shape:
        img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))
        print("Second image resized to match the first image.")

    # Calculate MSE
    mse = mean_squared_error(img1, img2)

    print("Mean Squared Error (MSE):", mse)

    # Display both images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img1, cmap="gray")
    plt.title("Image 1")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(img2, cmap="gray")
    plt.title("Image 2")
    plt.axis("off")

    plt.show()
```
